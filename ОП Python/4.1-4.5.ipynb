```python
print("Вычисление площади треугольника. Группы элементов: Основание и высота - 1, Две стороны и угол между ними - 2, Три стороны - 3")

while True:
    group = int(input("Введите номер группы элементов (0 для выхода): "))
    if group == 0:
        print("Вычисление прекращено") #Или print(False)
        break
    if group == 1:
        a, h = map(float, input("Введите основание и высоту: ").split())
        print(f"Площадь треугольника: {0.5 * a * h:.2f} м^2")
        #Контрольные числа 4 3 == 0.5 * 4 * 3 == 6.00
    elif group == 2:
        storona_1, storona_2, ygol = map(float, input("Введите две стороны и угол (в радианах) между ними: ").split())
        print(f"Площадь треугольника: {0.5 * storona_1 * storona_2 * sin(ygol):.2f} м^2")
        #Контрольные числа: 3 4 1.047 == 0.5 * 3 * 4 * 1.047 (Должно выводить 5.20) где 1.047 получается из угола 60 градусов: 60 * (3.14 / 180)
    elif group == 3:
        a, b, c = map(float, input("Введите три стороны через пробел: ").split())
        p = (a + b + c) / 2 #Полупериметр
        print(f"Площадь треугольника: {sqrt(p * (p - a) * (p - b) * (p - c)):.2f} м^2") #По формуле Герона
        #Контрольные числа: 3 4 5 == p = 12 / 2 => sqrt(6 * (6 - 3) * (6 - 4) * (6 - 5))) (Должно выводить 6.00)
    else:
        print("Ошибка! Введите номер группы от 0 до 3")
```

```python
N = int(input("Введите натуральное число N: "))
S = 0
for n in range(N + 1):
    form = ((-1) ** n) * (1 / (2 ** n)) #Формула для последовательности
    S += form
print(f"S = {S:.5f}")
```

### Контрольные числа 3; 4; 5 == 1 - 1 / 2 + 1 / 4 - 1 / 8 (при n = 3: 0.625) + 1 / 16 (при n = 4: 0.6875) - 1 / 32 (при n = 5: 0.65625)

```python
e = float(input("Введите значение эпсилон (например, 0.01): "))
total_sum = 0
n = 1
max_iterations = 100  # Защита от зацикливания

while n <= max_iterations:
    a_n = ((-1)**(n - 1)) / (n**n) # Формула члена ряда
    
    if abs(a_n) < e:
        break
        
    total_sum += a_n
    n += 1
else: # В случае если произошло зацикливание
    print("Достигнут лимит итераций! Расчет остановлен (зацикливание).")
    exit()

print(f"Сумма членов ряда = {total_sum:.10f}")
print(f"Количество членов ряда = {n-1}")
```

### При e == 0.01: n1 == 1; n2 == (-1)^1/4 = -0.25; n3 == (-1)^2/27 = 0.037037; n4 == (-1)^3/256 = -0.003906 (n4 не входит при эпсилон равном 0.01)
### Сумма == 1 - 0.25 + 0.037037 = 0.787037

```python
a, b, h = map(float, input("Введите начало, конец и шаг через пробел: ").split())

results = []
current_x = a

# Расчет значений в цикле
# Добавляем h/10 к верхней границе, чтобы из-за погрешности не потерять точку b
while current_x <= b + (h / 10):
    f_x = current_x - sin(current_x)
    
    # Сохраняем значения в список для таблицы
    results.append([round(current_x, 4), round(f_x, 6)])
    
    current_x += h

# Вывод готовой таблицы
headers = ["x", "F(x) = x - sin(x)"]
print("\nТаблица значений функции:")
print(tabulate(results, headers=headers, tablefmt="grid"))
```

### При значениях a = 0; b = 2; h = 0,5
### x = 0.0: == 0 - sin(0), вывод - 0.0
### x = 0.5: == 0.5 - sin(0.5), вывод - 0.0205744614 (калькулятор)
### x = 1.0: == 1 - sin(1), вывод - 0.158529015 (калькулятор)
### x = 1.5: == 1.5 - sin(1.5), вывод - 0.502505013 (калькулятор)
### x = 2.0: == 2 - sin(2), вывод - 1.09070257 (калькулятор)

```python
while True:
    try:
        rinput = input("Введите числа через пробел (для дробных используйте точку): ").strip()
        
        # Проверка на пустой ввод
        if not rinput:
            print("Ошибка: Вы ничего не ввели. Попробуйте еще раз.")
            continue
        
        # Пытаемся преобразовать всё в float
        numbers = [float(x) for x in rinput.split()]
        
        # Если всё прошло успешно, считаем сумму и выходим из цикла
        total_sum = sum(numbers)
        print(f"\nУспешно! В последовательности {len(numbers)} элементов.")
        print(f"Сумма введенной последовательности: {total_sum}")
        break
        
    except ValueError:
        print("Ошибка: Обнаружен некорректный символ (буква или знак). Вводите только числа через пробел.")
```
