### Упражнение №1:
### Проилюстрируйте графически сходящийся итерационный процесс при условии, что: -1 < f'(x) < 0 на отрезке [a; b]

```python
from numpy import *
import matplotlib.pyplot as plt
```

```python
def f(x): return 0.5 * cos(x) + 1

x = linspace(0, 2, 100)
plt.plot(x, f(x), label='f(x)')
plt.plot(x, x, label='y=x')
plt.grid(True, alpha=0.3)

x0 = 0.5
x_points = [x0]
for i in range(6):
    x_points.append(f(x_points[-1]))
    plt.plot([x_points[i], x_points[i]], [x_points[i], x_points[i+1]])
    plt.plot([x_points[i], x_points[i+1]], [x_points[i+1], x_points[i+1]])
    plt.plot(x_points[i], f(x_points[i]), 'ro', markersize=4)

plt.title('Метод простой итерации')
plt.xlabel('x'), plt.ylabel('y')
plt.legend(), plt.axis([0, 2, 0, 2]), plt.gca().set_aspect('equal')
plt.show()

print("Итерации:", [f"{i}: {val:.6f}" for i, val in enumerate(x_points)])
```

### Упражнение №2:
### Уточнить наименьший по модулю отличный от нуля корень уравнения x - sin(x) - 1 = 0 методом простой итерации с точностью до 10^-5

```python
print("Решение уравнения x * sin(x) - 1 = 0")
print("Метод простой итерации, точность 10^-5")

def fi(x):
    return x - 0.5 * (x * sin(x) - 1)

x = 1.1
print(f"x0 = {x}\n")
print(" n        x_n         |x_n - x_n-1|")
print("-" * 40)

for n in range(1, 20):
    x_new = fi(x)
    diff = abs(x_new - x)
    print(f"{n:2}    {x_new:.8f}    {diff:.8f}")
    
    if diff < 10 ** -5:
        print("-" * 40)
        print(f"\nКорень: x = {x_new:.8f}")
        print(f"Проверка: {x_new:.8f}·sin({x_new:.8f}) - 1 = {x_new * sin(x_new) - 1:.2}")
        break
    x = x_new
```
